```python


def count_words_in_file(filename):
    try:
        with open(filename, 'r') as file:
            content = file.read()
            
            words = content.split()
            return len(words)
    except FileNotFoundError:
        return "Error: File not found."
    except Exception as e:
        return f"An error occurred: {e}"


filename = "test.txt"


with open(filename, 'w') as f:
    f.write("virtat kohli is the best cricket player ")
    
print(f"The file '{filename}' has {count_words_in_file(filename)} words.")
```
