```python
 #Read/Write Employee Records (File Handling & CSV)

import csv
import os

filename = "employees.csv"

def write_employees():
    
    employees = [
        {"ID": "101", "Name": "sachin", "Department": "HR"},
        {"ID": "102", "Name": "virat ", "Department": "Engineering"},
        {"ID": "103", "Name": "rohit", "Department": "Sales"}
    ]

    
    with open(filename, mode='w', newline='') as file:
        
        fieldnames = ["ID", "Name", "Department"]
        writer = csv.DictWriter(file, fieldnames=fieldnames)

        writer.writeheader() 
        writer.writerows(employees) 
        print(f"Success: Data written to {filename}")

def read_employees():
    if not os.path.exists(filename):
        print("File does not exist. Please write data first.")
        return

    print(f"\n--- Reading Data from {filename} ---")
    
    with open(filename, mode='r') as file:
        reader = csv.DictReader(file)
        for row in reader:
            print(f"ID: {row['ID']} | Name: {row['Name']} | Dept: {row['Department']}")

if __name__ == "__main__":
    write_employees()
    read_employees()
```
