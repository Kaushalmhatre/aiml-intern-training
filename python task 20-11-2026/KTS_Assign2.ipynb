```python


def find_duplicates(input_list):
    
    seen = set()
    duplicates = set()
    
    for item in input_list:
        if item in seen:
            duplicates.add(item)
        else:
            seen.add(item)
            
    return list(duplicates)


my_list = [1, 2, 3, 2, 4, 5, 5, 5, 6, 1]
print(f"Original list: {my_list}")
print(f"Duplicate values: {find_duplicates(my_list)}")
```
