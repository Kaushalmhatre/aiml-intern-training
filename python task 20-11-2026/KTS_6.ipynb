```python
# 1. Reverse a string
def reverse_string(s):
    return s[::-1]

# 2. Check if a number is prime
def is_prime(n):
    if n <= 1: return False
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0: return False
    return True

# 3. Find the largest number in a list
def find_largest(lst):
    return max(lst) # Or use a loop

# 4. Count the frequency of words in a string (using dictionaries)
def word_frequency(text):
    words = text.lower().split()
    freq = {}
    for word in words:
        freq[word] = freq.get(word, 0) + 1
    return freq

# 5. Remove duplicates from a list (using sets)
def remove_duplicates(lst):
    return list(set(lst))

# 6. Check if a string is a palindrome
def is_palindrome(s):
    s = s.lower().replace(" ", "")
    return s == s[::-1]

# 7. Generate a Fibonacci sequence up to n terms
def fibonacci(n):
    seq = []
    a, b = 0, 1
    for _ in range(n):
        seq.append(a)
        a, b = b, a + b
    return seq

# 8. Find the factorial of a number using a loop
def factorial(n):
    result = 1
    for i in range(1, n + 1):
        result *= i
    return result

# 9. Sort a list of tuples based on the second element
def sort_tuples(tuples_list):
    # Sorts based on the element at index 1 (the second item)
    return sorted(tuples_list, key=lambda x: x[1])

# 10. Read a file and count the number of lines
def count_lines(filename):
    try:
        with open(filename, 'r') as file:
            return len(file.readlines())
    except FileNotFoundError:
        return "File not found"


print("1. Reverse 'hello':", reverse_string("hello"))
print("2. Is 17 prime?:", is_prime(17))
print("3. Largest in [3, 7, 2, 9]:", find_largest([3, 7, 2, 9]))
print("4. Word freq 'apple banana apple':", word_frequency("apple banana apple"))
print("5. Remove duplicates [1, 2, 2, 3, 4, 4]:", remove_duplicates([1, 2, 2, 3, 4, 4]))
print("6. Is 'Racecar' a palindrome?:", is_palindrome("Racecar"))
print("7. First 5 Fibonacci:", fibonacci(5))
print("8. Factorial of 5:", factorial(5))
print("9. Sort tuples [(1, 3), (2, 1), (3, 2)]:", sort_tuples([(1, 3), (2, 1), (3, 2)]))
```
