```python
#program 5   Convert Celsius to Fahrenheit.
```

```python
celsius = float(input("Enter temperature in Celsius: "))
fahrenheit = (celsius * 9/5) + 32
print(f"{celsius}°C is equal to {fahrenheit}°F.")
```

```python
#program 8  Swap two variables.
```

```python
num_a = input("Enter the value for number  A: ")
num_b = input("Enter the value for number B: ")
print(f"Before swap: A = {num_a}, B = {num_b}")


num_a, num_b = num_b, num_a

print(f"After swap:  A = {num_a}, B = {num_b}")
```

```python
#program 9  Calculate the average of three input numbers.
```

```python
n1 = float(input("Enter the first number: "))
n2 = float(input("Enter the second number: "))
n3 = float(input("Enter the third number: "))
average = (n1 + n2 + n3) / 3
print(f"The average of the three numbers is: {average}")
```
