```python

students = {}

def add_student(name, marks):
    students[name] = marks
    print(f"Added {name} with marks: {marks}")

def calculate_average():
    if not students:
        return 0
    total_marks = sum(students.values())
    return total_marks / len(students)

def display_students():
    print("\n--- Student Records ---")
    for name, marks in students.items():
        print(f"Name: {name} | Marks: {marks}")


add_student("kaushal", 85)
add_student("vivek", 92)
add_student("Chintan", 78)

display_students()
print(f"\nClass Average: {calculate_average():.2f}")
```
